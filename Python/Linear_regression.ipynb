```python
data = {"hours" : [0,1,2,3,4], "scores" : [55,60,65,70,75]}
```

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# we used scatter plot to find the relationship between hours and scores
sns.scatterplot(x="hours", y="scores", data=df)
```

```python
# linear regression model used the equation of a straight line (y=mx + b) to find the relationship between hours and scores

'''x = independent variable (hours), hours 
y - target variable (scores), scores, Dependent variable
m = slope of the line
b = y-intercept'''

x = df[["hours"]]
y = df["scores"]    

from sklearn.linear_model import LinearRegression
```

```python
#instantiate the linear regression model 
my_model = LinearRegression()
```

```python
my_model.fit(x,y)
```

```python
my_model.predict([[6]])
```
