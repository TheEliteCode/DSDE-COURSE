```python
def countdown(i):
    if i == 0:
        print("done!")
    else:
        print(i)
        countdown(i - 1)

countdown(5)
```

```python
def factorial(n):
    if n == 1:
        return n
    else:
        return n * factorial(n - 1)
```

```python
factorial(5) == 120
```

```python
def factorial (n):
    if n == 1:
        return n
    else:
        print("n = ",n, "; now calling factorial(n-1)", sep = "")
        lower_fact = factorial(n - 1)
        current_fact = n * lower_fact
        print("n =",n, "; factorial(n-1) returned ", lower_fact,"; multiplied by n to get ", current_fact, sep = "")
        return current_fact
```

```python
factorial(5)
```

```python
def factorial(n):
    result = 1
    count = 2

    while count <= n:
        result = result * count
        count = count + 1

    return result
```

```python
factorial(4)
```
